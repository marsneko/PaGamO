    - number of different difficulty
    - each combination as a one-hot coding (should be limit in length 7~8 ? ,which is 2^7 = 128 )

```python
from typing import Callable


def generate_data(df:pd.DataFrame,
                  pre_test_interval:tuple[int,int]|int,
                  treatment_interval:tuple[int,int]|int,
                  after_test: tuple[int,int]|int,
                  translate_treatment_func: Callable[[list|int],None] | None ):
    for qid in range(12):
        df[f'q_{qid + 1}'] = df['question_id'].apply(lambda x: x[qid] if qid < len(x) else None)
        df[f'q_{qid + 1}_corr'] = df['is_correct'].apply(lambda x: int(x[qid]) if qid < len(x) else None)
        df[f'q_{qid + 1}_diff'] = df['difficulty_level'].apply(lambda x: int(x[qid]) if qid < len(x) else None)

    if isinstance(pre_test_interval,tuple):
        qfst = pre_test_interval[0] - 1
        qfin = pre_test_interval[1] - 1
        df['pre_test_corr'] = df['is_correct'].apply(lambda x: x[qfst:qfin+1] if qfin < len(x) else None)
        df['pre_test_diff'] = df['difficulty_level'].apply(lambda x: x[qfst:qfin+1] if qfin < len(x) else None)
    else:
        qidex = pre_test_interval - 1
        df['pre_test_corr'] = df["is_correct"].apply(lambda x : x[qidex] if qidex < len(x) else None)
        df['pre_test_diff'] = df['difficulty_level'].apply(lambda x : x[qidex] if qidex < len(x) else None)

    if isinstance(treatment_interval, tuple):
        qfst = treatment_interval[0] - 1
        qfin = treatment_interval[1] - 1
        df['treatment_corr'] = df['is_correct'].apply(lambda x:x[qfst:qfin+1] if qfin < len(x) else None)
        df['treatment_diff'] = df['difficulty_level'].apply(lambda x: x[qfst:qfin+1] if qfin < len(x) else None)
    else:
        qidex = treatment_interval - 1
        df['treatment_corr'] = df["is_correct"].apply(lambda x : x[qidex] if qidex < len(x) else None)
        df['treatment_diff'] = df['difficulty_level'].apply(lambda x : x[qidex] if qidex < len(x) else None)

    if isinstance(after_test,tuple):
        qfst = after_test[0] - 1
        qfin = after_test[1] - 1
        df['after_test_corr'] = df['is_correct'].apply(lambda x:x[qfst:qfin+1] if qfin < len(x) else None)
        df['after_test_diff'] = df['difficulty_level'].apply(lambda x: x[qfst:qfin+1] if qfin < len(x) else None)
    else:
        qidex = after_test - 1
        df['after_test_corr'] = df["is_correct"].apply(lambda x : x[qidex] if qidex < len(x) else None)
        df['after_test_diff'] = df['difficulty_level'].apply(lambda x : x[qidex] if qidex < len(x) else None)

    if translate_treatment_func is not None:
        df['treatment'] = df['treatment_diff'].apply(lambda x : translate_treatment_func(x))

    return df
```

```python
treatment_one_hot_coding = {
    (i, j, k): int(idx)
    for idx, (i, j, k) in enumerate(
        [(a, b, c) for a in range(1, 4) for b in range(1, 4) for c in range(1, 4)],
        start=1
        )
    }

def translate_func_count(lis):
    if lis is None:
        return None
    cates = [1,2,3]
    nums = [ lis.count(lev) for lev in cates]
    return nums

def treatment_one_hot_coding_func(input):
    if input is None:
        return None
    input = tuple(input)
    return treatment_one_hot_coding.get(input,None)
```

```python
def OLS_estimate(df:pd.DataFrame, dep_var:str, indep_var:list):
    xstr = ""
    for var in indep_var:
        xstr += f"{var} +"
    xstr = xstr[:-1]

    formula = f"{dep_var} ~ {xstr}"
    model = smf.ols(data = df, formula= formula).fit()
    print(model.summary())
    return model
```

```python
def logit_estimate(df:pd.DataFrame, dep_var:str, indep_var:list):
    xstr = ""
    for var in indep_var:
        xstr += f"{var} +"
    xstr = xstr[:-1]

    formula = f"{dep_var} ~ {xstr}"
    model = smf.logit(data = df, formula= formula).fit()
    print(model.summary())
    return model
```

```python
THREADHOLD_EASY_TOOLONG = 60 * 6
THREADHOLD_HARD_TOOSHORT = 60 * 0.27
```

```python
regdfs = [regdf.query(f"q_5_diff == {i}").copy() for i in [1,2,3]]
treatment = [f"num_diff_{i}" for i in [2,3]]
conti_indep_var = ["C(q_1_corr) : C(q_1_diff)",'q5_time_cost','treatment_time_cost*C(is_lazy)']
conti_indep_var.extend(treatment)
for df,dif in zip( regdfs,[1,2,3]):
    print(len(df.index))
    df['treatment_time_cost'] = df['time_diff'].apply(lambda x : np.sum(x[1:4])/60 if len(x) > 3 else None)
    df['q5_time_cost'] = df['time_diff'].apply(lambda x : x[4]/60 if len(x) > 4 else None)
    df['is_lazy_easy'] = df.apply(lambda row : 1 if len(row['time_diff']) > 4 and
                                            (
                                               (row['q_2_diff'] == 1 and row['time_diff'][1]>=THREADHOLD_EASY_TOOLONG)
                                            or (row['q_3_diff'] == 1 and row['time_diff'][2]>=THREADHOLD_EASY_TOOLONG)
                                            or (row['q_4_diff'] == 1 and row['time_diff'][3]>=THREADHOLD_EASY_TOOLONG)
                                            )
                                            else 0
                             ,axis = 1
                             )
    df['is_lazy_hard'] = df.apply(lambda row : 1 if len(row['time_diff']) > 4 and
                                            (
                                               (row['q_2_diff'] == 3 and row['time_diff'][1]<=THREADHOLD_HARD_TOOSHORT)
                                            or (row['q_3_diff'] == 3 and row['time_diff'][2]<=THREADHOLD_HARD_TOOSHORT)
                                            or (row['q_4_diff'] == 3 and row['time_diff'][3]<=THREADHOLD_HARD_TOOSHORT)
                                            )
                                            else 0
                             ,axis = 1
                             )
    df['is_lazy'] = df.apply(lambda row: 1 if row['is_lazy_easy'] == 1 or row['is_lazy_hard'] == 1 else 0,axis = 1)
    print(f"\n\n difficulty of question 5 == {dif} \n\n")
    OLS_estimate(df, dep_var="q_5_corr", indep_var=conti_indep_var)
```
